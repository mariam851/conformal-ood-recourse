# 11 · Generate Final Paper Tables and Figures

Reads the raw CSVs and exports LaTeX/CSV tables to `results/tables/` and PNG/PDF figures to `results/figures/`.

```python
from conformal_trees.evaluation.paper_artifacts import make_all
from conformal_trees.utils import resolve_path

made = make_all(
    bench_csv=resolve_path("results/raw/recourse_benchmark.csv"),
    gate_csv=resolve_path("results/raw/gate_benchmark.csv"),
    tables_dir=resolve_path("results/tables"),
    figures_dir=resolve_path("results/figures"),
    headline_target=0.6,
)

for name, files in made.items():
    print(f"{name:34s} -> {', '.join(f.name for f in files)}")
```
