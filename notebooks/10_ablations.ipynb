# 10 · Ablations: search budget and action space

Comparing `gen10` (10 generations), `gen100` (100 generations), and `relax` (allowing `marital-status` and `relationship` to change) at target = 0.6 across 3 seeds.

```python
import pandas as pd
from conformal_trees.evaluation.paper_artifacts import (
    load_benchmark,
    paired_effects,
    summary_strings,
    summary_tables,
)
from conformal_trees.utils import resolve_path

pd.set_option("display.max_columns", 50)
df = load_benchmark(resolve_path("results/raw/ablations.csv"))
mean, sd = summary_tables(df)
display(summary_strings(mean, sd))
```

```python
for tag in sorted(df.variant.unique()):
    print(f"=== Paired effect (gate_IF - none) | Variant: {tag} ===")
    eff = paired_effects(
        df, ["gate_IF"], [0.6],
        ["flag_IF", "flag_LOF", "flag_kNN", "cost", "n_changed", "robust"],
        variant=tag
    )
    display(eff[["metric", "diff", "ci_low", "ci_high", "n_pairs", "p_holm"]].round(4))
```
